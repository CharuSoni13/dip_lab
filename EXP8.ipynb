```python
# ==========================================================
# Image Restoration Techniques
# Mean Filter, Gaussian Filter and Wiener Filter
# ==========================================================

# Install required libraries
!pip install opencv-python scipy

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from scipy.signal import wiener

# ==========================================================
# STEP 1: Upload Image at Runtime
# ==========================================================

uploaded = files.upload()

# Get uploaded image name automatically
image_name = list(uploaded.keys())[0]

# Read image in grayscale
img = cv2.imread(image_name, 0)

# ==========================================================
# STEP 2: Add Noise to Image
# ==========================================================

# Create Gaussian noise
noise = np.random.normal(0, 25, img.shape)

# Add noise to image
noisy_img = img + noise

# Clip values between 0 and 255
noisy_img = np.clip(noisy_img, 0, 255).astype(np.uint8)

# ==========================================================
# STEP 3: Mean Filter Restoration
# ==========================================================

mean_restored = cv2.blur(noisy_img, (5,5))

# ==========================================================
# STEP 4: Gaussian Filter Restoration
# ==========================================================

gaussian_restored = cv2.GaussianBlur(noisy_img, (5,5), 0)

# ==========================================================
# STEP 5: Wiener Filter Restoration
# ==========================================================

wiener_restored = wiener(noisy_img, (5,5))

# ==========================================================
# STEP 6: Display Results
# ==========================================================

plt.figure(figsize=(15,10))

# Original Image
plt.subplot(2,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

# Noisy Image
plt.subplot(2,2,2)
plt.imshow(noisy_img, cmap='gray')
plt.title("Noisy Image")
plt.axis("off")

# Mean Filter Output
plt.subplot(2,2,3)
plt.imshow(mean_restored, cmap='gray')
plt.title("Mean Filter Restored Image")
plt.axis("off")

# Gaussian Filter Output
plt.subplot(2,2,4)
plt.imshow(gaussian_restored, cmap='gray')
plt.title("Gaussian Filter Restored Image")
plt.axis("off")

plt.tight_layout()
plt.show()

# ==========================================================
# STEP 7: Display Wiener Filter Separately
# ==========================================================

plt.figure(figsize=(5,5))
plt.imshow(wiener_restored, cmap='gray')
plt.title("Wiener Filter Restored Image")
plt.axis("off")
plt.show()
```
