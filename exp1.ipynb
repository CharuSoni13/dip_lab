```python
#AIM: Create a program to demonstrate Geometric transformations- Image rotation, scaling, and translation.

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Read uploaded image
image_name = list(uploaded.keys())[0]
img = cv2.imread(image_name)

# Convert BGR to RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Get image dimensions
rows, cols = img.shape[:2]

# ============================================
# 1. Translation
# ============================================

# Translation matrix
tx = 100   # shift right
ty = 50    # shift down

translation_matrix = np.float32([
    [1, 0, tx],
    [0, 1, ty]
])

translated_img = cv2.warpAffine(
    img,
    translation_matrix,
    (cols, rows)
)

# ============================================
# 2. Scaling
# ============================================

scaled_img = cv2.resize(
    img,
    None,
    fx=1.5,    # scale x-axis
    fy=1.5     # scale y-axis
)

# ============================================
# 3. Rotation
# ============================================

# Rotation around center
center = (cols // 2, rows // 2)

rotation_matrix = cv2.getRotationMatrix2D(
    center,
    45,      # angle
    1.0      # scale
)

rotated_img = cv2.warpAffine(
    img,
    rotation_matrix,
    (cols, rows)
)

# ============================================
# Display Results
# ============================================

plt.figure(figsize=(15,10))

# Original Image
plt.subplot(2,2,1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

# Translated Image
plt.subplot(2,2,2)
plt.imshow(translated_img)
plt.title("Translated Image")
plt.axis("off")

# Scaled Image
plt.subplot(2,2,3)
plt.imshow(scaled_img)
plt.title("Scaled Image")
plt.axis("off")

# Rotated Image
plt.subplot(2,2,4)
plt.imshow(rotated_img)
plt.title("Rotated Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
