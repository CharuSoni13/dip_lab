```python
# ==========================================================
# Mean, Standard Deviation and Correlation Coefficient
# ==========================================================

# Install OpenCV
!pip install opencv-python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# ==========================================================
# STEP 1: Upload Image at Runtime
# ==========================================================

uploaded = files.upload()

# Get uploaded image name automatically
image_name = list(uploaded.keys())[0]

# Read image in grayscale
img = cv2.imread(image_name, 0)

# ==========================================================
# STEP 2: Display Original Image
# ==========================================================

plt.figure(figsize=(5,5))
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")
plt.show()

# ==========================================================
# STEP 3: Compute Mean
# ==========================================================

mean_value = np.mean(img)

# ==========================================================
# STEP 4: Compute Standard Deviation
# ==========================================================

std_deviation = np.std(img)

# ==========================================================
# STEP 5: Compute Correlation Coefficient
# ==========================================================

# Take neighboring pixels
x = img[:, :-1].flatten()   # all columns except last
y = img[:, 1:].flatten()    # all columns except first

# Correlation coefficient
correlation_matrix = np.corrcoef(x, y)

correlation_coefficient = correlation_matrix[0,1]

# ==========================================================
# STEP 6: Display Results
# ==========================================================

print("===================================")
print("Image Statistical Parameters")
print("===================================")

print(f"Mean of Image                : {mean_value:.2f}")

print(f"Standard Deviation           : {std_deviation:.2f}")

print(f"Correlation Coefficient      : {correlation_coefficient:.4f}")

print("===================================")
```
